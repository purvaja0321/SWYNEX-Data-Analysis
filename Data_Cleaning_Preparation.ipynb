```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
print("Rows after removing duplicates:", len(df_clean))
```

```python
text_columns = ["InvoiceNo", "StockCode", "Description", "Country"]

for col in text_columns:
    df_clean[col] = df_clean[col].astype("string").str.strip()
```

```python
df_clean = df_clean.dropna(
    subset=["Description", "CustomerID"]
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean = df_clean[df_clean["Quantity"] > 0]
```

```python
print("Invalid quantities remaining:",
      (df_clean["Quantity"] <= 0).sum())
```

```python
df_clean = df_clean[df_clean["UnitPrice"] > 0]
```

```python
print("Invalid prices remaining:",
      (df_clean["UnitPrice"] <= 0).sum())
```

```python
df_clean["InvoiceNo"] = df_clean["InvoiceNo"].astype("string")
df_clean["StockCode"] = df_clean["StockCode"].astype("string")
df_clean["Description"] = df_clean["Description"].astype("string")
df_clean["Country"] = df_clean["Country"].astype("string")
```

```python
df_clean["CustomerID"] = pd.to_numeric(
    df_clean["CustomerID"],
    errors="coerce"
).astype("Int64")
```

```python
df_clean["Quantity"] = pd.to_numeric(
    df_clean["Quantity"],
    errors="coerce"
).astype("Int64")
```

```python
df_clean["UnitPrice"] = pd.to_numeric(
    df_clean["UnitPrice"],
    errors="coerce"
)
```

```python
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    errors="coerce"
)
```

```python
df_clean["TotalAmount"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
).round(2)
```
